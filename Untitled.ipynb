```python
df_test.drop(columns=['Cabin', 'Ticket', 'Name'], axis=1, inplace=True, errors='ignore')
```

```python
df_test['Age'] = df_test['Age'].fillna(df_test['Age'].median())
df_test['Embarked'] = df_test['Embarked'].fillna(df_test['Embarked'].mode()[0])
```

```python
df_train.info()
```

```python
women = df_train.loc[df_train.Sex == 'female']["Survived"]
rate_women = sum(women)/len(women)

print("% of women who survived:", rate_women)
```

```python

men = df_train.loc[df_train.Sex == 'male']["Survived"]
rate_men = sum(men)/len(men)

print("% of men who survived:", rate_men)
```

```python

# Countplot for survival
sns.countplot(data=df_train, x='Survived')
plt.title("Survival Count")
plt.show()

# Histogram for Age
df_train['Age'].hist(bins=20, edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python

# Survival by Gender
sns.countplot(data=df_train, x='Survived', hue='Sex')
plt.title("Survival by Gender")
plt.show()

# Boxplot of Fare by Survival
sns.boxplot(data=df_train, x='Survived', y='Fare')
plt.title("Fare Distribution by Survival")
plt.show()
```

'''Multivariate Analysis:'''

```python
# Select only numeric columns from the DataFrame
numeric_data = df_train.select_dtypes(include=['number'])

# Compute the correlation matrix
corr_matrix = numeric_data.corr()

# Plot the heatmap
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python

# Group survival rate by Pclass and Gender
survival_rate = df_train.groupby(['Pclass', 'Sex'])['Survived'].mean().reset_index()
print(survival_rate)
```

```python
from sklearn.ensemble import RandomForestClassifier

y = df_train["Survived"]

features = ["Pclass", "Sex", "SibSp", "Parch"]
X = pd.get_dummies(df_train[features])
X_test = pd.get_dummies(df_test[features])

model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
model.fit(X, y)
predictions = model.predict(X_test)
output = pd.DataFrame({'PassengerId': df_test.PassengerId, 'Survived': predictions})
output.to_csv('submission.csv', index=False)
print("Your submission was successfully saved!")
```
